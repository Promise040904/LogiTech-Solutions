```python
import pickle

model = pickle.load(
    open("trained_model.pkl","rb")
)

distance = 180
traffic = 0
weather = 1
fuel = 30

prediction = model.predict([
    [distance,
     traffic,
     weather,
     fuel]
])

if prediction[0] == 1:
    print("Delivery Delay Expected")
else:
    print("Delivery On Time")
```
